```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
data = """
lstm networks are powerful for processing sequential data. 
recurrent neural networks have limitations with long sequences.
lstm cells contain memory gates that help preserve information.
the forget gate decides what information to discard from memory.
the input gate controls what new information enters the memory.
the output gate determines what information leaves the memory.
vanishing gradient problem affects standard recurrent networks.
lstm networks solve the vanishing gradient problem effectively.
gru networks are simplified versions of lstm networks.
bidirectional lstm networks process sequences in both directions.
attention mechanisms improve the performance of neural networks.
transformer models use self attention instead of recurrence.
deep learning has revolutionized natural language processing.
convolutional networks excel at image recognition tasks.
dropout regularization prevents overfitting in neural networks.
batch normalization improves training stability and speed.
activation functions introduce non linearity to networks.
relu activation is widely used in hidden layers.
softmax activation provides probability distributions for classification.
gradient descent optimizes the weights of neural networks.
adam optimizer adapts learning rates for each parameter.
learning rate decay helps convergence in deep learning.
early stopping prevents overfitting by monitoring validation loss.
""".lower().strip()
```

```python
data
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([data])
total_words = len(tokenizer.word_index) + 1
```

```python
input_sequences = []
for line in data.split('\n'):
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)
```

```python
max_sequence_len = max([len(x) for x in input_sequences])
```

```python
input_sequences = np.array(pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre'))
```

```python
tokenizer.word_counts
```

```python
tokenizer.word_index
```

```python
X = input_sequences[:, :-1]
y = input_sequences[:, -1]
```

```python
early_stop = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss', patience=3, restore_best_weights=True)

model_rnn = tf.keras.Sequential([
    tf.keras.layers.Embedding(total_words, 100, input_length=max_sequence_len-1),
    tf.keras.layers.SimpleRNN(150, return_sequences=True),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.SimpleRNN(150),
    tf.keras.layers.Dense(total_words, activation='softmax')
])
```

```python
model_rnn.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
history_rnn = model_rnn.fit(X, y, epochs=100, verbose=1, validation_split=0.1, callbacks=[early_stop])
```

```python
seed_text = "lstm networks are"
for _ in range(5):
    token_list = tokenizer.texts_to_sequences([seed_text])[0]
    token_list = pad_sequences([token_list], maxlen=max_sequence_len-1, padding='pre')
    predictions = model_rnn.predict(token_list, verbose=0)[0]
    
    # Use temperature sampling for diversity instead of just argmax
    temperature = 0.7
    predictions = predictions ** (1 / temperature)
    predictions = predictions / np.sum(predictions)
    predicted_idx = np.random.choice(len(predictions), p=predictions)
    
    for word, index in tokenizer.word_index.items():
        if index == predicted_idx:
            seed_text += " " + word
            break
print("RNN Result:", seed_text)
```

```python
model_lstm = tf.keras.Sequential([
    tf.keras.layers.Embedding(total_words, 100, input_length=max_sequence_len-1),
    tf.keras.layers.LSTM(200, return_sequences=True, dropout=0.2, recurrent_dropout=0.2),
    tf.keras.layers.LSTM(200, dropout=0.2, recurrent_dropout=0.2),
    tf.keras.layers.Dense(total_words, activation='softmax')
])

model_lstm.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
history_lstm = model_lstm.fit(X, y, epochs=100, verbose=1, validation_split=0.1, callbacks=[early_stop])
```

```python
seed_text = "lstm networks are"
for _ in range(5):
    token_list = tokenizer.texts_to_sequences([seed_text])[0]
    token_list = pad_sequences([token_list], maxlen=max_sequence_len-1, padding='pre')
    predictions = model_lstm.predict(token_list, verbose=0)[0]
    
    # Use temperature sampling for diversity instead of just argmax
    temperature = 0.7
    predictions = predictions ** (1 / temperature)
    predictions = predictions / np.sum(predictions)
    predicted_idx = np.random.choice(len(predictions), p=predictions)
    
    for word, index in tokenizer.word_index.items():
        if index == predicted_idx:
            seed_text += " " + word
            break
print("LSTM Result:", seed_text)
```
